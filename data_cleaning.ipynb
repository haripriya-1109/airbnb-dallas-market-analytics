```python
clean_df = df.copy()
```

```python
clean_df = clean_df.drop(columns=["neighbourhood_group", "license"])
```

```python
clean_df = clean_df.dropna(subset=["price"])
```

```python
clean_df["reviews_per_month"] = clean_df["reviews_per_month"].fillna(0)
```

```python
clean_df["last_review"] = pd.to_datetime(clean_df["last_review"])
```

```python
print("Original rows:", len(df))
print("Cleaned rows:", len(clean_df))
print("Original columns:", len(df.columns))
print("Cleaned columns:", len(clean_df.columns))
```

```python
clean_df["minimum_nights"] = clean_df["minimum_nights"].fillna(
    clean_df["minimum_nights"].median()
)
```

```python
price_99th = clean_df["price"].quantile(0.99)

clean_df["is_price_outlier"] = (
    clean_df["price"] > price_99th
)
```

```python
print("99th percentile price:", price_99th)
print("Price outliers:", clean_df["is_price_outlier"].sum())
```

```python
clean_df.isnull().sum().sort_values(ascending=False)
```

```python
clean_df["is_price_outlier"].value_counts()
```

```python
clean_df = clean_df.drop(
    columns=["host_profile_id", "host_name"]
)
```

```python
print("Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))
print(clean_df.columns.tolist())
```

```python
clean_df["price_capped"] = clean_df["price"].clip(
    upper=price_99th
)
```

```python
print("Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))
print("\nColumns:")
print(clean_df.columns.tolist())

print("\nMissing values:")
print(clean_df.isnull().sum().sort_values(ascending=False))

print("\nPrice statistics:")
print(clean_df[["price", "price_capped"]].describe())
```

```python
clean_df["host_id"] = clean_df["host_id"].astype("Int64")

clean_df["calculated_host_listings_count"] = (
    clean_df["calculated_host_listings_count"].astype("Int64")
)

clean_df["minimum_nights"] = clean_df["minimum_nights"].astype("Int64")

clean_df.to_csv("airbnb_dallas_clean.csv", index=False)
```
